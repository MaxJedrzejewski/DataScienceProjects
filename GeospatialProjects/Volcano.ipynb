```python
# Imports
import folium
from folium.plugins import HeatMap
import pandas as pd
import geopandas as gpd
```

```python

# Load the CSV files
url = 'https://gist.githubusercontent.com/mattkram/9684863843254402942dfede27af2cb7/raw/2590dd8185b833aacf247c0595edbb07a025a6d7/Smithsonian_VOTW_Holocene_Volcanoes.csv'
filepath = "C:/Users/maxje/OneDrive/Dokumenty/anaconda_projects/tectonicplates-master/GeoJSON/PB2002_boundaries.json"
volcanos = pd.read_csv(url)
tectonic_boundaries = gpd.read_file(filepath)
# Show the first 5 rows
tectonic_boundaries.head()
```

```python
# Transform geometry column to coordinates
tectonic_boundaries['coordinates'] = tectonic_boundaries.apply(lambda x: [(b,a) for (a,b) in list(x.geometry.coords)], axis='columns')
tectonic_boundaries.drop('geometry', axis=1, inplace=True)
```

```python
# Create a folium map centered around a specific location
map_center = [0, 0]  # Center of the map (latitude, longitude)
volcano_map = folium.Map(location=map_center, zoom_start=2)

# Add markers for each volcano
for index, row in volcanos.iterrows():
    folium.Marker(
        location=[row['Latitude'], row['Longitude']],
        tooltip=row['Volcano_Name'],
        popup="<strong>Name: </strong>" + row['Volcano_Name'] + "<br><strong>Volcano type: </strong>" + row['Primary_Volcano_Type'],
        max_width="100%",
        icon=folium.Icon(color='red')
    ).add_to(volcano_map)

# Save the map to an HTML file
volcano_map.save('volcano_map.html')

# Display the map
volcano_map
```

```python
# Create base map with tectonic plate boundaries
volcano_heatmap = folium.Map(location=map_center, zoom_start=2)

for i in range(len(tectonic_boundaries)):
    folium.PolyLine(locations=tectonic_boundaries.coordinates.iloc[i], weight=2, color='black').add_to(volcano_heatmap)



# Create heatmap of volcanos
HeatMap(data=volcanos[['Latitude','Longitude']], radius=20).add_to(volcano_heatmap)

# Display the map
volcano_heatmap
```

```python
volcano_heatmap = folium.Map(location=map_center, zoom_start=2)
tectonic_plates = gpd.read_file(filepath)

# Add GeoJson layer for tectonic boundaries
folium.GeoJson(
    tectonic_plates,
    tooltip=folium.GeoJsonTooltip(fields=['Name'], aliases=['Boundary Name:']),
    color='black'
).add_to(volcano_heatmap)

# Create heatmap of volcanos
HeatMap(data=volcanos[['Latitude','Longitude']], radius=20).add_to(volcano_heatmap)

volcano_heatmap
```
